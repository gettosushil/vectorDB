```python
from pinecone import Pinecone, ServerlessSpec 
```

```python
pc = Pinecone(api_key="pcsk_dummy_replace_with_your_key")
```

```python
pc.list_indexes()
```

```python
pc.create_index(name="test2", 
               dimension = 3, 
               spec=ServerlessSpec(
        cloud="aws",
        region="us-east-1"
    )  ) # free tier use only us-east-1
```

```python
idx = pc.Index("test2")
```

```python
data = [('vec1', [1.0, 2.0, 3.0]), ('vec2', [11.0, 12.0, 13.0]), ('vec3', [31.0, 32.0, 33.0])]
```

```python
idx.upsert(vectors = data)
```

```python
idx.upsert(vectors = [
    ("A", [0.1, 0.1, 0.1]),
("B", [0.2, 0.2, 0.2]),
("C", [0.3, 0.3, 0.3]),
("D", [0.4, 0.4, 0.4]),
("E", [0.5, 0.5, 0.5])
])
```

```python
idx.update(id="B", values=[0.2, 0.2, 0.55])
```

```python
idx.query(vector = [0.1, 0.1 , 0.1], top_k=5, include_values = True)
```

```python
idx.delete(ids = ['A', 'B'])
```

```python
idx.fetch(ids=["D", "E"])
```
